# Search Capability Analysis
To fulfill a research request like "Azure OpenAI and SQL Server", the agent relies on several data sources. Here is the current status:

| Source | Capability | Configuration |
| :--- | :--- | :--- |
| **Microsoft Graph** | Real-time SharePoint/OneDrive Search | ✅ Active |
| **Azure AI Search** | Indexed private document collections | ✅ Configured |
| **Local Cache** | Previously crawled/processed docs | ✅ Active |
| **Web Index** | Background indexed external websites | ⚠️ Crawl-based only |

### 💡 Recommendation for Research
The agent does **not** have a "live" search engine (like Bing API) for arbitrary web queries. It only searches the specific URLs listed in your configuration.

**To get the best research results:**
1. **Upload relevant whitepapers or PDFs** directly to the chat.
2. **Add research URLs** (e.g., `learn.microsoft.com`) to your `EXTERNAL_WEB_SOURCES` in your environment file.
3. **Use the pipe operator `|`** if you want to search multiple specific sub-topics at once (e.g., `azure openai features | sql server vector support`).
